```python
import os
import matplotlib.pyplot as plt
import numpy as np
```

```python
filenames = [
    "SZA_aerosol_r_0.2.dat",
    "SZA_clouds_0.2.dat",
    "SZA_clouds_0.3.dat",
    "SZA_clouds_0.15.dat",
    "SZA_clouds_0.1.dat",
    "SZA_clouds_0.05.dat",
]

labels = [
    "Aerosol (Rural; TBAER = 0.2)",
    "Clouds (ZCLOUD = 2.5, 6.0 km; TCLOUD = 0.2; NRE = 8)",
    "Clouds (ZCLOUD = 2.5, 6.0 km; TCLOUD = 0.3; NRE = 8)",
    "Clouds (ZCLOUD = 2.5, 6.0 km; TCLOUD = 0.15; NRE = 8)",
    "Clouds (ZCLOUD = 2.5, 6.0 km; TCLOUD = 0.1; NRE = 8)",
    "Clouds (ZCLOUD = 2.5, 6.0 km; TCLOUD = 0.05; NRE = 8)",
]

colors = [
    'blue',
    'red', 
    'green', 
    'purple', 
    'slateblue', 
    'rebeccapurple',
    'indigo'
]

folder_path = 'C:\\Users\\PC\\Documents\\_PRACTIQUES_UDG\\2_SBDART\\SBDART\\_MARTI\\Dades'
data_files = [
    f"{folder_path}\\{filename}" for filename in filenames
]
```

```python
# Gràfic
plt.figure(figsize=(10, 6))
for data_file in data_files:
    if os.path.exists(data_file):
        data = np.loadtxt(data_file)
        if data.shape[1] == 2: # Comprova que té dues columnes
            idx = data_files.index(data_file)
            plt.plot(data[:, 0], data[:, 1], marker='o', linestyle='-', label=labels[idx], color=colors[idx % len(colors)])
        else:
            print(f"Data file {data_file} does not have the expected format.")
    else:
        print(f"Data file {data_file} does not exist.")
plt.title('Clarity Index vs Diffuse Fraction for different SZA')
plt.xlabel('Clarity Index (KT = GLOB / EXT)')
plt.ylabel('Diffuse Fraction (DF = DIF / GLOB)')
plt.grid(True)
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.legend()
# plt.tight_layout()
plt.xticks(np.arange(0, 1.1, 0.1))
plt.yticks(np.arange(0, 1.1, 0.1))
plt.show()
```
