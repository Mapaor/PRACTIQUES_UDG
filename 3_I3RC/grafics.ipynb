```python
import os
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Updated to use the new clean datasets
filenames = [
    "aerosol_KT_DF.dat",  # Rural aerosol (τ=0.2)
    "cloud_KT_DF.dat"     # Thin cloud (τ=0.2)
]

labels = [
    "Rural Aerosol (τ=0.2, reff=0.5μm)",
    "Thin Cloud (τ=0.2, reff=10μm)"
]

colors = [
    'blue',    # Aerosol
    'red'      # Cloud
]

# Updated to use the main directory where the clean data files are located
folder_path = 'C:\\Users\\PC\\Documents\\GitHub\\I3RC\\Dades'
data_files = [
    f"{folder_path}\\{filename}" for filename in filenames
]
```

```python
# Plot the I3RC Monte Carlo results
plt.figure(figsize=(10, 8))

for data_file in data_files:
    if os.path.exists(data_file):
        data = np.loadtxt(data_file)
        if data.shape[1] == 2:  # Check that it has two columns (KT, DF)
            idx = data_files.index(data_file)
            plt.plot(data[:, 0], data[:, 1], marker='o', linestyle='-', 
                    linewidth=2, markersize=6, label=labels[idx], color=colors[idx])
        else:
            print(f"Data file {data_file} does not have the expected format.")
    else:
        print(f"Data file {data_file} does not exist.")

plt.title('I3RC Monte Carlo Results: Clearness Index vs Diffuse Fraction\nAerosols vs Clouds (τ=0.2)', fontsize=14)
plt.xlabel('KT (Global / Ext)', fontsize=12)
plt.ylabel('DF (Diff / Global)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.xlim(0, 1.1)
plt.ylim(0, 1.1)
plt.legend(fontsize=11)
plt.xticks(np.arange(0, 1.1, 0.1))
plt.yticks(np.arange(0, 1.1, 0.1))

plt.tight_layout()
plt.show()
```
